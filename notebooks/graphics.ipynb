```python
import ast

import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Настройка графиков
sns.set_style("whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)

df = pd.read_csv(r"..\datasets\all_q_True.csv")

# ============================================================
# 2. Первичный осмотр
# ============================================================
print("=" * 50)
print("Форма датасета:", df.shape)
print("\nИнформация о столбцах:")
print(df.info())
print("\nПервые 5 строк:")
print(df.head())

# ============================================================
# 3. Пропуски
# ============================================================
print("\n" + "=" * 50)
print("Пропущенные значения:")
missing = df.isnull().sum()
missing_pct = (missing / len(df)) * 100
missing_df = pd.DataFrame({"Количество": missing, "Процент": missing_pct})
print(missing_df[missing_df["Количество"] > 0])

# Если context полностью пустой, можно его удалить или проигнорировать
if df["context"].isnull().all():
    print("\nСтолбец 'context' состоит только из NaN.")

# ============================================================
# 4. Анализ числовых признаков (здесь только question_id)
# ============================================================
print("\n" + "=" * 50)
print("Описательная статистика для question_id:")
print(df["question_id"].describe())

# Проверка уникальности question_id
unique_ids = df["question_id"].nunique()
print(f"Уникальных question_id: {unique_ids} из {len(df)} (дубликатов: {len(df) - unique_ids})")

# ============================================================
# 5. Анализ категориальных признаков
# ============================================================
print("\n" + "=" * 50)
print("Распределение dataset:")
print(df["dataset"].value_counts())

print("\nРаспределение context_used:")
print(df["context_used"].value_counts(normalize=True) * 100)

# Визуализация
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.countplot(data=df, x="dataset", ax=axes[0], order=df["dataset"].value_counts().index)
axes[0].set_title("Распределение dataset")
sns.countplot(data=df, x="context_used", ax=axes[1])
axes[1].set_title("Распределение context_used")
plt.tight_layout()
plt.show()

# ============================================================
# 6. Анализ текстовых полей: question и correct_answer
# ============================================================
# Длина вопроса в символах и словах
df["question_len"] = df["question"].astype(str).apply(len)
df["question_words"] = df["question"].astype(str).apply(lambda x: len(x.split()))

print("\n" + "=" * 50)
print("Статистика длины вопроса (символы):")
print(df["question_len"].describe())
print("\nСтатистика длины вопроса (слова):")
print(df["question_words"].describe())

# Гистограммы
fig, axes = plt.subplots(1, 2, figsize=(14, 5))
sns.histplot(df["question_len"], bins=30, ax=axes[0], kde=True)
axes[0].set_title("Длина вопроса (символы)")
sns.histplot(df["question_words"], bins=30, ax=axes[1], kde=True)
axes[1].set_title("Длина вопроса (слова)")
plt.tight_layout()
plt.show()


# ============================================================
# 7. Обработка correct_answer: str или list
# ============================================================
# Функция для безопасного парсинга строки-списка
def parse_correct_answer(value):
    if isinstance(value, list):
        return value, len(value)
    if isinstance(value, str):
        # Если строка похожа на список (начинается с '[')
        if value.strip().startswith("["):
            try:
                parsed = ast.literal_eval(value)
                if isinstance(parsed, list):
                    return parsed, len(parsed)
            except:
                pass
        return value, 1  # строка-ответ, не список
    return value, np.nan


# Применяем парсинг
parsed_answers = df["correct_answer"].apply(parse_correct_answer)
df["answer_parsed"] = parsed_answers.apply(lambda x: x[0])
df["answer_count"] = parsed_answers.apply(lambda x: x[1])

# Анализ типа ответа
df["answer_type"] = df["answer_parsed"].apply(lambda x: "list" if isinstance(x, list) else "str")
print("\n" + "=" * 50)
print("Типы correct_answer после парсинга:")
print(df["answer_type"].value_counts())

# Если есть списки – распределение количества элементов в списке
if df["answer_type"].eq("list").any():
    print("\nРаспределение числа элементов в списке (answer_count):")
    print(df.loc[df["answer_type"] == "list", "answer_count"].value_counts().sort_index())
    plt.figure()
    sns.countplot(data=df[df["answer_type"] == "list"], x="answer_count")
    plt.title("Количество правильных ответов (для случаев со списком)")
    plt.show()

# Статистика длины строкового представления correct_answer
df["answer_str_len"] = df["correct_answer"].astype(str).apply(len)
print("\nСтатистика длины строки correct_answer:")
print(df["answer_str_len"].describe())
plt.figure()
sns.histplot(df["answer_str_len"], bins=30, kde=True)
plt.title("Длина correct_answer (строковое представление)")
plt.show()

# ============================================================
# 8. Дополнительные проверки
# ============================================================
# Дубликаты по всему датафрейму
dup_rows = (
    df[
        [
            "dataset",
            "question_id",
            "question",
            "context",
            "correct_answer",
            "context_used",
            "question_len",
            "question_words",
            "answer_count",
            "answer_type",
            "answer_str_len",
        ]
    ]
    .duplicated()
    .sum()
)
print("\n" + "=" * 50)
print(f"Полных дубликатов строк: {dup_rows}")

# Дубликаты вопросов
dup_questions = df["question"].duplicated().sum()
print(f"Дубликатов вопросов: {dup_questions}")

# Проверка связи dataset и context_used
print("\nКросс-таблица dataset vs context_used:")
print(pd.crosstab(df["dataset"], df["context_used"], normalize="index") * 100)

# ============================================================
# 9. Сохранение обработанного датафрейма (опционально)
# ============================================================
# df.to_csv("processed_dataset.csv", index=False)
```

```python
# Предполагаем, что df уже загружен и содержит колонки:
# 'dataset' (3 уникальных значения), 'question', 'context'
# Если контекст отсутствует – соответствующие значения игнорируются

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# Проверяем, что датасетов ровно три
datasets_list = df["dataset"].unique()
if len(datasets_list) != 3:
    print(f"Ожидается 3 датасета, но обнаружено {len(datasets_list)}: {list(datasets_list)}")
else:
    # Вычисление длин
    df["question_len"] = df["question"].astype(str).apply(len)
    # Контекст может быть NaN, сохраняем NaN
    df["context_len"] = df["context"].apply(lambda x: len(str(x)) if pd.notnull(x) else np.nan)

    # Создаём один рисунок с двумя подграфиками (вопрос и контекст)
    fig, axes = plt.subplots(1, 2, figsize=(14, 6))

    # Гистограммы длины вопроса
    for ds in datasets_list:
        subset = df[df["dataset"] == ds]
        axes[0].hist(subset["question_len"], bins=40, alpha=0.5, label=ds)
    axes[0].set_title("Длина вопроса (символы)")
    axes[0].set_xlabel("Количество символов")
    axes[0].set_ylabel("Частота")
    axes[0].legend()

    # Гистограммы длины контекста (только для непустых значений)
    for ds in datasets_list:
        subset = df[df["dataset"] == ds]
        context_vals = subset["context_len"].dropna()
        if len(context_vals) > 0:
            axes[1].hist(context_vals, bins=100, alpha=0.5, label=ds)
    axes[1].set_title("Длина контекста (символы)")
    axes[1].set_xlabel("Количество символов")
    axes[1].set_ylabel("Частота")
    axes[1].legend()

    plt.tight_layout()
    plt.show()
```
